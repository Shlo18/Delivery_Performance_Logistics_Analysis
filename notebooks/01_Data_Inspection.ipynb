```python
df = pd.read_csv(
    "../data/Apl_Logistics.csv",
    encoding="latin1"
)
print("Dataset loaded:",df.shape)
```

```python
df_clean = df.copy()
print("Clean dataset created:",df_clean.shape)
```

```python
df_clean["Days for shipping (real)"]=pd.to_numeric(
    df_clean["Days for shipping (real)"],
    errors="coerce"
)
df_clean["Days for shipment (scheduled)"]=pd.to_numeric(
    df_clean["Days for shipment (scheduled)"],
    errors="coerce"
)
df_clean["Late_delivery_risk"]=pd.to_numeric(
    df_clean["Late_delivery_risk"],
    errors="coerce"
)
```

```python
print(
    df_clean[
        [
            "Days for shipping (real)",
            "Days for shipment (scheduled)",
            "Late_delivery_risk"
								]
				].dtypes
)
```

```python
df_clean[
    [
        "Days for shipping (real)",
        "Days for shipment (scheduled)",
        "Late_delivery_risk"
				]
].isnull().sum()
```

```python
text_columns =[
    "Market",
    "Order Region",
    "Order Country",
    "Shipping Mode",
    "Customer Segment",
    "Delivery Status",
    "Order Status"
]

for col in text_columns:
    df_clean[col]=df_clean[col].astype("string").str.strip()
```

```python
print("Markets:")
print(df_clean["Market"].unique())
print("/nShipping Modes:")
print(df_clean["Shipping Mode"].unique())
```

```python
print("/nRegions:")
print(df_clean["Order Region"].unique())
```

```python
print(
    df_clean["Late_delivery_risk"].value_counts(
        dropna=False
				)
)
```

```python
df_clean["Delay_Gap"]=(
    df_clean["Days for shipping (real)"]
   - df_clean["Days for shipment (scheduled)"]
)
```

```python
df_clean[
    [
        "Days for shipping (real)",
        "Days for shipment (scheduled)",
        "Delay_Gap"
				]
].head(10)
```

```python
df_clean["Delivery_performance"]=np.select(
    [
        df_clean["Delay_Gap"]<0,
        df_clean["Delay_Gap"]==0,
        df_clean["Delay_Gap"]>0
				],
                [
                    "Early",
                    "On_Time",
                    "Delayed"
																],
                                                                default="Unkonwn"
)
```

```python
df_clean["Delivery_performance"].value_counts()
```

```python
pd.crosstab(
    df_clean["Delivery_performance"],
    df_clean["Late_delivery_risk"]
)
```

```python
print("Final rows:",len(df_clean))
print("Final columns:",len(df_clean.columns))
print("\nMissing critical values:")
print(
    df_clean[critical_columns].isnull().sum()
)
print("\nDuplicates:")
print(df_clean.duplicated().sum())
print("\nDelivery performance:")
print(df_clean["Delivery_performance"].value_counts())
print("\nDelay Gap:")
print(df_clean["Delay_Gap"].describe())
```

```python
df_clean.to_csv(
    "../outputs/cleaned_data/APL_Logistics_Cleaned.csv",
    index=False
)
print("Cleaned dataset saved successffully!")
```
